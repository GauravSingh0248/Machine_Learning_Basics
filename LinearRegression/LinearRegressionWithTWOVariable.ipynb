```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression
# plt.style.use("seaborn")
```

```python
# Step -01 Generate Data
X,y=make_regression(n_samples=500,n_features=10,n_informative=5,noise=25.8,random_state=0)
print(X.shape,y.shape)
n_features=X.shape[1]
```

```python
pd.DataFrame(X).head() #head(n=100)
```

```python
# Step --02 Normalize the Data 
# X.mean(axis=0).shape
def normalise(X):
    u=X.mean(axis=0)
    std=X.std(axis=0)
    return (X-u)/std
    
```

```python
X=normalise(X)
pd.DataFrame(X).head()
```

```python
# Step--03 Visualise y as a function of each feature 
# means every feature vs y plot

for f in range(0,10):
    plt.subplot(4,3,f+1)
    plt.scatter(X[:,f],y)
plt.show()
```

```python
# Step --4 Train Test Split
XT,Xt,yT,yt=train_test_split(X,y,test_size=0.3,shuffle=False,random_state=0)
```

```python
print(XT.shape,yT.shape)
print(Xt.shape,yt.shape)
```

```python
# Step--05 Modelling / Linear Regression with Multiple Features

def preprocess(X):
    # add a column of 1s in X as 0th column
    if X.shape[1]==n_features: # just for safety
        m=X.shape[0]
        ones=np.ones((m,1))
        X=np.hstack((ones,X))
    return X
```

```python
def hypothesis(X,theta):
    return np.dot(X,theta) # fast / vectorized way of writing the code ---faster than loop 

def loss(X,y,theta):
    yp=hypothesis(X,theta)
    error=np.mean((y-yp)**2)
    return error

def gradient(X,y,theta):
    yp=hypothesis(X,theta)
    grad=np.dot(X.T,(yp-y))
    m=X.shape[0]
    return grad/m

def train(X,y,learning_rate=0.1,max_iters=100):
    n=X.shape[1]
    theta=np.random.randn(n)
    error_list=[]
    
    for i in range(max_iters):
        e=loss(X,y,theta)
        error_list.append(e)
        grad=gradient(X,y,theta)
        theta=theta-learning_rate*grad

    plt.plot(error_list)
    return theta
    
```

```python
theta=train(XT,yT)
print(theta)
```

```python
pd.DataFrame(X).head()
```

```python
Xt=preprocess(Xt)
XT=preprocess(XT)
print(XT.shape,Xt.shape)
```

```python
def r2Score(y,yp):
    ymean=y.mean()
    num=np.sum((y-yp)**2)
    denom=np.sum((y-ymean)**2)
    return 1-num/denom
```

```python
yp=hypothesis(Xt,theta)
```

```python
r2Score(yt,yp)
```
